```python
import numpy as np
import pandas as pd

def numeros_impares(lista):
  lista_impar = []
  for i in lista:
    if i % 2 != 0:
      lista_impar.append(i)
  return lista_impar

def numeros_primos(lista):
  lista_primos = []
  for i in lista:
    if i > 1:
      for item in range(2,i):
        if item % i == 0:
          break
      else:
        lista_primos.append(i)
  return lista_primos

def listas(lista1, lista2):
  lista_unica = []
  for item in lista1:
    if item not in lista2:
      lista_unica.append(item)

  for item in lista2:
    if item not in lista1:
      lista_unica.append(item)
  return lista_unica

def segundo_maior(lista):
  return sorted(set(lista), reverse=True)[1]

def ordenar_por_nome(lista):
  return sorted(lista, key=lambda pessoa:pessoa[0])

def identificar_tratar_outliers(valor):
  valores = np.array(valor)
  primeiro_quartil = np.percentile(valores, 25)
  terceiro_quartil = np.percentile(valores,75)
  iqr = terceiro_quartil - primeiro_quartil

  limite_inferior = primeiro_quartil - 1.5 * iqr
  limite_superior = terceiro_quartil + 1.5 * iqr
  # Todos os valores que estiverem fora desses limites (superiores e inferiores, são considerados outliers)
  outliers = [x for x in valores if x < limite_inferior or x > limite_superior]
  # para tratar um outlier, primeiramente precisamos analise se faz sentido ou não manter aquele valor... caso não faça sentido:
  valor_tratado = [x for x in valores if limite_inferior <= x <= limite_superior]
  return {
      'outliers' : outliers,
      'tratados' : valor_tratado
  }

def concatenar_dataframes(lista_df, eixo=0):
  return pd.concat(lista_df, axis=eixo, ignore_index=(eixo==0))

def df_csv(arquivo):
  """
  para realizar a leitura de um arquivo CSV basta utilizarmos o .read_csv da bliblioteca pandas,
  para exibir as primeiras linhas utilizamos o .head()
  link dataset: https://www.kaggle.com/datasets/abdulwadood11220/youtube-data-for-analytics-600-rows?resource=download
  """
  df = pd.read_csv(arquivo)
  return df.head()

def filtro(arquivo):
  df = pd.read_csv(arquivo)
  """
  para realizar filtros no DataFrame, podemo utilizar tanto o loc quando o iloc...
  para esse exercicio eu utilizei o .loc para realizar uma condição
  """
  filtro = df.loc[df['comment_count'] > 1000]
  return filtro

def lidar_com_valores_ausentes(arquivo):

  """
  para realizar a ultima atividade sobre valores ausentes, irei utilizar o seguinte dataset:
  https://www.kaggle.com/datasets/brendan45774/test-file
  """

  df = pd.read_csv(arquivo)
  print("valores ausentes: ")
  print(df.isna().sum())
  """
  para modelos de ML, valores ausentes são super prejudiciais, pois podem ocorrer problemas e até mesmo levar falhas nos modelos.
  por isso é bom remover logo de cara ou, caso tenhamos poucos valores, substituir pela média ou mediana (tambem dependendo do tamanho do dataset)

  para esse exemplo eu preferi remover
  """
  df = df.dropna()
  print("\nsem valores ausentes")
  print(df.isna().sum())
  return df
```
